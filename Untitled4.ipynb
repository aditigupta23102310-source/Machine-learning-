```python
#1d_numpy array
import numpy as np
array_1d=np.array([1,2,3,4,5])
print(array_1d)
#2d_ numpy array
import numpy as np
array_2d=np.array([[1,2,3,4],[2,3,5,6]])
print(array_2d)
#3d_numpy array
import numpy as np
array_3d=np.array([[[1,2,3],[4,5,6],[6,7,8]]])
print(array_3d)
#data type
import numpy as np
a=np.array([1,2,3],dtype=complex)
print(a)
#np.arange
import numpy as np
a=np.arange(1,10)
print(a)
#with reshape
import numpy as np
a=np.arange(1,13).reshape(6,2)
print(a)
#np.ones and np.zeroes(used to initialize value)
import numpy as np
a=np.zeros((3,4))
print(a)
import numpy as np
a=np.ones((3,3))
print(a)
#np.random
import numpy as np
a=np.random.random((2,7))
print(a)
#np.linearly space
import numpy as np
a=np.linspace(-10,10,10,complex)
print(a)
#np.identity
import numpy as np
a=np.identity(3,dtype=int)
print(a)
```

```python
#array attributes
import numpy as np
a1=np.arange(10)
a2=np.arange(12,dtype=float).reshape(3,4)
a3=np.arange(8).reshape(2,2,2)
print(a1)
print(a2)
print(a3)
#ndim(dimension batata hai)
print(a1.ndim)
print(a2.ndim)
print(a3.ndim)
#shape
print(a1.shape)
print(a2.shape)
print(a3.shape)
#size
print(a1.size)
print(a2.size)
print(a3.size)
#itemsize(int,float size)
print(a1.itemsize)
print(a2.itemsize)
print(a3.itemsize)
#if item size of integer is 32 bit waise 64 bit leta hai by default
a1=np.arange(8,dtype=np.int32)
print(a1.itemsize)
#astype
import numpy as np
a3=np.arange(8).reshape(2,2,2)
print(a3.astype(np.int32))

```

```python

import numpy as np
a1=np.arange(12).reshape(3,4)
a2=np.arange(12,24).reshape(3,4)
print(a1)
print(a2)
#ARITHMETIC OPERATION
#scalar operator
print(2+a1)
print(2-a1)
print(2*a1)
print(a1/2)
print(a1**2)
#relational operator
print(a1==15)
print(a2>10)
#VECTOR OPERATION
#arithmetic operator
print(a1+a2)
print(a1-a2)
print(a1*a2)
print(a1/a2)
#relational operator
print(a1==a2)
print(a1>a2)
```

```python
#Array Function
import numpy as np
a1=np.arange(12,24).reshape(3,4)
#max/min/sum/prod
print(np.max(a1))
print(np.min(a1))
print(np.sum(a1))
print(np.prod(a1))
#0->col and 1->row
print(np.max(a1,axis=0))
print(np.min(a1,axis=0))
print(np.sum(a1,axis=0))
print(np.prod(a1,axis=0))
#mean/median/std/var
print(np.mean(a1))
print(np.median(a1))
print(np.std(a1))
print(np.var(a1))
#for a column
print(np.mean(a1,axis=0))
print(np.median(a1,axis=0))
print(np.std(a1,axis=0))
print(np.var(a1,axis=0))
#trignometric function
print(np.sin(a1))
print(np.cos(a1))
print(np.tan(a1))
#dot product
a2=np.arange(12,24).reshape(3,4)
a3=np.arange(1,13).reshape(4,3)
print(np.dot(a2,a3))
#log and exponent
print(np.log(a1))
print(np.exp(a1))
#round/floor/ceil
a1=np.random.random((2,3))
print(np.round(a1))
print(np.floor(a1))
print(np.ceil(a1))





```
